```python
import shutil
import random
import os

IMG_DIR = "data/images"
LBL_DIR = "data/labels"

os.makedirs("dataset/images/train", exist_ok=True)
os.makedirs("dataset/images/val", exist_ok=True)
os.makedirs("dataset/labels/train", exist_ok=True)
os.makedirs("dataset/labels/val", exist_ok=True)

images = [f for f in os.listdir(IMG_DIR) if f.endswith((".jpg", ".png", ".jpeg"))]
random.shuffle(images)

split = int(0.8 * len(images))
train_images = images[:split]
val_images = images[split:]

for img in train_images:
    shutil.move(f"{IMG_DIR}/{img}", f"dataset/images/train/{img}")
    shutil.move(
        f"{LBL_DIR}/{os.path.splitext(img)[0]}.txt",
        f"dataset/labels/train/{os.path.splitext(img)[0]}.txt"
    )

for img in val_images:
    shutil.move(f"{IMG_DIR}/{img}", f"dataset/images/val/{img}")
    shutil.move(
        f"{LBL_DIR}/{os.path.splitext(img)[0]}.txt",
        f"dataset/labels/val/{os.path.splitext(img)[0]}.txt"
    )

print("✅ Dataset split completed")


```

```python
from ultralytics import YOLO

model = YOLO("yolov8n.pt")
print("YOLOv8 loaded ✅")
```

```python
model.train(
    data="data.yaml",
    epochs=15,
    imgsz=640,
    batch=8
)
```

```python
model = YOLO("runs/detect/train/weights/best.pt")
print("Custom pothole model loaded ✅")
```

```python
results = model("omkar.jpg", conf=0.25)
results[0].show()
```

```python
import cv2
import numpy as np

img = cv2.imread("omkar.jpg")
h, w, _ = img.shape
img_area = h * w

report = []

for box in results[0].boxes:
    x1, y1, x2, y2 = map(int, box.xyxy[0])
    area = (x2 - x1) * (y2 - y1)
    ratio = area / img_area

    if ratio < 0.02:
        severity = "Low"
    elif ratio < 0.07:
        severity = "Moderate"
    else:
        severity = "Severe"

    report.append({
        "issue": "Pothole",
        "severity": severity,
        "confidence": round(float(box.conf), 2),
        "area_percent": round(ratio * 100, 2)
    })

report
```

```python
overlay = img.copy()

for item, box in zip(report, results[0].boxes):
    x1, y1, x2, y2 = map(int, box.xyxy[0])
    
    color = (0,255,0) if item["severity"]=="Low" else \
            (0,255,255) if item["severity"]=="Moderate" else \
            (0,0,255)

    cv2.rectangle(overlay, (x1,y1), (x2,y2), color, 3)
    cv2.putText(
        overlay,
        item["severity"],
        (x1, y1-10),
        cv2.FONT_HERSHEY_SIMPLEX,
        0.9,
        color,
        2
    )

import matplotlib.pyplot as plt
plt.imshow(cv2.cvtColor(overlay, cv2.COLOR_BGR2RGB))
plt.axis("off")
```
