```python
def image_chat_pipeline(image_path, user_reply=None):

    # First turn (image uploaded)
    if user_reply is None:
        image_description = analyze_image(image_path)
        issue_type = classify_image_issue(image_description)

        question = QUESTION_FLOW[issue_type][0]

        record = {
            "time": str(datetime.now()),
            "image": image_path,
            "image_description": image_description,
            "issue_type": issue_type,
            "step": 0,
            "bot_question": question
        }

        conversation_data.append(record)
        save_to_json()
        return question

    # Next turns (user replies)
    else:
        record = conversation_data[-1]
        record["user_answer"] = user_reply
        record["step"] += 1

        step = record["step"]
        questions = QUESTION_FLOW[record["issue_type"]]

        if step < len(questions):
            next_question = questions[step]
            record["bot_question"] = next_question
            save_to_json()
            return next_question
        else:
            record["status"] = "completed"
            save_to_json()
            return "Thank you. Your report has been recorded."
```

```python
q1 = image_chat_pipeline("asphalt-damage-road-cracking.jpg")
print("Bot:", q1)

q2 = image_chat_pipeline(
    "asphalt-damage-road-cracking.jpg",
    user_reply="Nagpur, near Suryoday College"
)
print("Bot:", q2)

q3 = image_chat_pipeline(
    "asphalt-damage-road-cracking.jpg",
    user_reply="Around 3 months"
)
print("Bot:", q3)
```
